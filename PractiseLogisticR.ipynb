## Importing Libararies

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

## Importing dataset

```python
dataset = pd.read_csv('/content/drive/MyDrive/ML Practise/Machine-Learning-A-Z-Codes-Datasets/Part 2 - Regression/Section 6 - Polynomial Regression/Python/Position_Salaries.csv')
dataset
```

```python
X = dataset.iloc[:, 1:-1].values
y = dataset.iloc[:, -1].values
```

## Build Linear Regression Model

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X, y)
```

## Train in Polynomial regression

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree = 4)
X_poly = poly_reg.fit_transform(X)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly, y)
```

## Visualize the Linear regression results

```python
plt.scatter(X,y, color='red')
plt.plot(X, lin_reg.predict(X), color = 'blue')
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

## Visualize the polynomial regression result

```python
plt.scatter(X,y, color='red')
plt.plot(X, lin_reg_2.predict(X_poly), color = 'blue')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('Position Level')
plt.ylabel('Salary')
plt.show()
```

## predicting the new results with linear regression

```python
lin_reg.predict([[6.5]])
```

## Predicting the new results with polynomial regression

```python
lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))
```
