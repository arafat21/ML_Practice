## Preprocessing and Cleaning


```python
# Positive review 1
dataset['rating']=dataset['rating'].apply(lambda x: 0 if x<3 else 1)
```

```python
dataset['rating'].value_counts()
```

```python
dataset['reviewText'] = dataset['reviewText'].str.lower()
```

```python
dataset
```

```python
import re
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords
from bs4 import BeautifulSoup
```

```python
## Removing Special Character
dataset['reviewText']=dataset['reviewText'].apply(lambda x:re.sub('[^a-z A-z 0-9-]+', '',x))
## Remove the stopswords
dataset['reviewText']=dataset['reviewText'].apply(lambda x:" ".join([y for y in x.split() if y not in stopwords.words('english')]))
## Remove url
dataset['reviewText']=dataset['reviewText'].apply(lambda x: re.sub(r'(http|https|ftp|ssh)://([\w_-]+(?:(?:\.[\w_-]+)+))([\w.,@?^=%&:/~+#-]*[\w@?^=%&/~+#-])?', '' , str(x)))
## Remove html tags
dataset['reviewText']=dataset['reviewText'].apply(lambda x: BeautifulSoup(x, 'lxml').get_text())
## Remove any additional spaces
dataset['reviewText']=dataset['reviewText'].apply(lambda x: " ".join(x.split()))
```

```python
dataset.head()
```

```python
## Lemmatizer
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
lemmatizer = WordNetLemmatizer()
```

```python
def lemmatize_words(text):
    return " ".join([lemmatizer.lemmatize(word) for word in text.split()])
```

```python
dataset['reviewText']=dataset['reviewText'].apply(lambda text: lemmatize_words(text))
```

```python
dataset.head()
```

## Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(dataset['reviewText'],dataset['rating'],
                                              test_size=0.20)
```

## Bag of Words

```python
from sklearn.feature_extraction.text import CountVectorizer
bow=CountVectorizer()
X_train_bow=bow.fit_transform(X_train).toarray()
X_test_bow=bow.transform(X_test).toarray()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf=TfidfVectorizer()
X_train_tfidf=tfidf.fit_transform(X_train).toarray()
X_test_tfidf=tfidf.transform(X_test).toarray()
```

```python
X_train_bow
```

## ML Model

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow = GaussianNB().fit(X_train_bow,y_train)
nb_model_tfidf = GaussianNB().fit(X_train_tfidf,y_train)
```

## Accuracy Score

```python
from sklearn.metrics import confusion_matrix,accuracy_score, classification_report
```

```python
y_pred_bow=nb_model_bow.predict(X_test_bow)
```

```python
y_pred_tfidf = nb_model_tfidf.predict(X_test_tfidf)
```

```python
print("BOW Accuracy: ",accuracy_score(y_test,y_pred_bow))
print("TFIDF Accuracy:",accuracy_score(y_test,y_pred_tfidf))
```

```python
confusion_matrix(y_test,y_pred_tfidf)
```

```python
confusion_matrix(y_test,y_pred_tfidf)
```

```python
print("TFIDF accuracy: ",accuracy_score(y_test,y_pred_tfidf))
```
